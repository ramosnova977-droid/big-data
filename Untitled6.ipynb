```python

import pandas as pd

with zipfile.ZipFile(ruta_zip, "r") as zip_ref:
    with zip_ref.open("train.csv") as archivo:
        df = pd.read_csv(archivo, nrows=10000)

print("Filas y columnas:", df.shape)
print("\nColumnas:")
print(df.columns.tolist())

display(df.head())
```

```python

# Análisis exploratorio inicial

import pandas as pd

resumen = pd.DataFrame({
    "Columna": df.columns,
    "Tipo de dato": df.dtypes.astype(str).values,
    "Valores nulos": df.isnull().sum().values,
    "Porcentaje nulos": (
        df.isnull().mean().values * 100
    ).round(2)
})

display(resumen)
```

```python

# VALIDACIÓN INICIAL DEL DATASET

print("1. Registros duplicados por ID:")
print(df["id"].duplicated().sum())

print("\n2. Estadísticas de toxicidad:")
print(df["target"].describe())

print("\n3. Distribución de rating:")
print(df["rating"].value_counts(dropna=False))

print("\n4. Rango de fechas:")

fechas = pd.to_datetime(
    df["created_date"],
    errors="coerce",
    utc=True
)

print("Fecha mínima:", fechas.min())
print("Fecha máxima:", fechas.max())
print("Fechas inválidas:", fechas.isna().sum())

print("\n5. Anotadores de toxicidad:")
print(df["toxicity_annotator_count"].describe())

print("\n6. Anotadores de identidad:")
print(df["identity_annotator_count"].describe())
```

```python

# Análisis de toxicidad y rating

import pandas as pd

# Clasificación propuesta, no etiqueta original
df["clasificacion_toxicidad"] = (
    df["target"] >= 0.5
).map({
    True: "Toxico",
    False: "No toxico"
})

# Distribución de comentarios
print("1. Clasificación de toxicidad")
print(df["clasificacion_toxicidad"].value_counts())

# Relación entre rating y toxicidad
print("\n2. Rating frente a toxicidad")
tabla = pd.crosstab(
    df["rating"],
    df["clasificacion_toxicidad"]
)
display(tabla)

# Porcentajes por rating
print("\n3. Porcentaje dentro de cada rating")
display(
    pd.crosstab(
        df["rating"],
        df["clasificacion_toxicidad"],
        normalize="index"
    ).mul(100).round(2)
)

# Promedio de puntuación según rating
print("\n4. Toxicidad promedio por rating")
display(
    df.groupby("rating")["target"]
    .agg(["count", "mean", "median", "max"])
)
```
